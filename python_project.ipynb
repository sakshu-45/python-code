```python
import json
import os
from datetime import date

FILE_NAME = "attendance.json"


def load_data():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as file:
            return json.load(file)
    return {}


def save_data(data):
    with open(FILE_NAME, "w") as file:
        json.dump(data, file, indent=4)


students = load_data()


def add_student():
    roll = input("Enter Roll Number: ")
    name = input("Enter Student Name: ")

    if roll in students:
        print("Student already exists!")
    else:
        students[roll] = {
            "name": name,
            "attendance": {}
        }
        save_data(students)
        print("Student added successfully!")


def view_students():
    if not students:
        print("No students found.")
        return

    print("\n--- Student List ---")

    for roll, details in students.items():
        print(f"Roll: {roll} | Name: {details['name']}")


def mark_attendance():
    roll = input("Enter Roll Number: ")

    if roll not in students:
        print("Student not found!")
        return

    status = input("Enter P for Present or A for Absent: ").upper()

    if status not in ["P", "A"]:
        print("Invalid attendance status!")
        return

    today = str(date.today())

    students[roll]["attendance"][today] = status

    save_data(students)

    print("Attendance marked successfully!")


def view_attendance():
    roll = input("Enter Roll Number: ")

    if roll not in students:
        print("Student not found!")
        return

    print("\nName:", students[roll]["name"])
    print("--- Attendance Records ---")

    records = students[roll]["attendance"]

    for day, status in records.items():
        print(day, "-", "Present" if status == "P" else "Absent")


def attendance_percentage():
    roll = input("Enter Roll Number: ")

    if roll not in students:
        print("Student not found!")
        return

    records = students[roll]["attendance"]

    total = len(records)
    present = list(records.values()).count("P")

    if total == 0:
        print("No attendance records found.")
        return

    percentage = (present / total) * 100

    print("Student:", students[roll]["name"])
    print("Total Classes:", total)
    print("Present:", present)
    print("Absent:", total - present)
    print(f"Attendance Percentage: {percentage:.2f}%")


# Main Menu

while True:

    print("\n===== STUDENT ATTENDANCE MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Mark Attendance")
    print("4. View Attendance")
    print("5. Attendance Percentage")
    print("6. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        view_students()

    elif choice == "3":
        mark_attendance()

    elif choice == "4":
        view_attendance()

    elif choice == "5":
        attendance_percentage()

    elif choice == "6":
        print("Thank you for using the system!")
        break

    else:
        print("Invalid choice! Please try again.")
```
